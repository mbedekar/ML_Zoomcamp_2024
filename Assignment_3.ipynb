```python
df_new = df.loc[:, df.columns != 'loan']
df_new.y = (df_new.y == 'yes').astype(int)

##df_new = df_new.drop('balance', axis=1)


df_full_train, df_test = train_test_split(df_new, test_size=0.2, random_state=42)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=42)


df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

y_train = df_train.y.values
y_val = df_val.y.values
y_test = df_test.y.values

del df_train['y']
del df_val['y']
del df_test['y']


categorical_columns = list(df_train.dtypes[df_train.dtypes == 'object'].index)
numeric_columns = list(df_train.dtypes[df_train.dtypes == 'int64'].index)

dv = DictVectorizer(sparse=False)

train_dict = df_train[categorical_columns + numeric_columns].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)

val_dict = df_val[categorical_columns + numeric_columns].to_dict(orient='records')
X_val = dv.transform(val_dict)

model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_val)
y_pred = model.predict_proba(X_val)[:, 1]
final_y = (y_pred >= 0.5)
accuracy_ori = (final_y == y_val).mean()
accuracy_ori
```
